```python
import random

def fitness(wolf):
    temp, fan, cooling = wolf

    comfort = (temp - 24)**2
    energy = 0.1 * (fan**2 + cooling**2)

    return comfort + energy


wolves = 10
iterations = 50

population = []

for i in range(wolves):
    population.append([
        random.uniform(18, 30),
        random.uniform(1, 5),
        random.uniform(1, 10)
    ])

for t in range(iterations):

    population.sort(key=fitness)

    alpha = population[0]
    beta = population[1]
    delta = population[2]

    a = 2 - (2 * t / iterations)

    for i in range(3, wolves):
        new_position = []

        for j in range(3):

            r1 = random.random()
            r2 = random.random()

            A1 = 2 * a * r1 - a
            C1 = 2 * r2

            D_alpha = abs(C1 * alpha[j] - population[i][j])
            X1 = alpha[j] - A1 * D_alpha

            r1 = random.random()
            r2 = random.random()

            A2 = 2 * a * r1 - a
            C2 = 2 * r2

            D_beta = abs(C2 * beta[j] - population[i][j])
            X2 = beta[j] - A2 * D_beta

            r1 = random.random()
            r2 = random.random()

            A3 = 2 * a * r1 - a
            C3 = 2 * r2

            D_delta = abs(C3 * delta[j] - population[i][j])
            X3 = delta[j] - A3 * D_delta

            new_value = (X1 + X2 + X3) / 3
            new_position.append(new_value)

        new_position[0] = max(18, min(30, new_position[0]))
        new_position[1] = max(1, min(5, new_position[1]))
        new_position[2] = max(1, min(10, new_position[2]))

        population[i] = new_position

population.sort(key=fitness)
best = population[0]

print("AIR CONDITIONER TUNING USING GWO")
print("Best Temperature:", round(best[0], 2), "C")
print("Best Fan Speed:", round(best[1], 2))
print("Best Cooling Power:", round(best[2], 2))
print("Minimum Fitness:", round(fitness(best), 2))
```
