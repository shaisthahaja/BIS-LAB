```python
import random

shades = [
    (180, 30, 60),    # Red
    (220, 100, 120),  # Pink
    (120, 20, 40),    # Maroon
    (200, 80, 50),    # Coral
    (150, 50, 100)    # Berry
]

population_size = 10
mutation_rate = 0.2
generations = 20

def create_shade():
    return random.choice(shades)

target = (200, 50, 80)

def fitness(shade):
    distance = sum(abs(shade[i] - target[i]) for i in range(3))
    return 765 - distance

def crossover(parent1, parent2):
    return tuple((parent1[i] + parent2[i]) // 2 for i in range(3))

def mutate(shade):
    if random.random() < mutation_rate:
        shade = tuple(
            max(0, min(255, c + random.randint(-20, 20)))
            for c in shade
        )
    return shade

population = [create_shade() for _ in range(population_size)]

for generation in range(generations):
    population.sort(key=fitness, reverse=True)

    new_population = population[:2]  # Keep best two shades

    while len(new_population) < population_size:
        parent1 = random.choice(population[:5])
        parent2 = random.choice(population[:5])

        child = crossover(parent1, parent2)
        child = mutate(child)

        new_population.append(child)

    population = new_population

best_shade = max(population, key=fitness)

print("Best lipstick shade (RGB):", best_shade)
print("Fitness score:", fitness(best_shade))


print("All Shades in the Population:")

for i, shade in enumerate(population, 1):
    print("Shade", i, ":", shade, "Fitness:", fitness(shade))
```
