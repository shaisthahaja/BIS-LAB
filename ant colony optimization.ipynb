```python
import random
import math

n = 50
partners = 20

ants = 20
iterations = 100
alpha = 1
beta = 2
rho = 0.5
Q = 100

random.seed(10)

cities = [(random.randint(1, 100), random.randint(1, 100))
          for i in range(n)]

depot = (50, 50)

def distance(a, b):
    return math.sqrt((a[0] - b[0]) ** 2 +
                     (a[1] - b[1]) ** 2)

dist = [[distance(cities[i], cities[j])
         for j in range(n)] for i in range(n)]

depot_dist = [distance(depot, cities[i]) for i in range(n)]

pheromone = [[1.0] * n for i in range(n)]
depot_pheromone = [1.0] * n

best_routes = None
best_distance = float("inf")

sizes = [3] * 10 + [2] * 10

for iteration in range(iterations):

    for ant in range(ants):

        unvisited = list(range(n))
        routes = []
        total = 0

        for size in sizes:

            route = []
            current = -1

            for step in range(size):

                choices = []
                weights = []

                for city in unvisited:

                    if current == -1:
                        d = depot_dist[city]
                        p = depot_pheromone[city]
                    else:
                        d = dist[current][city]
                        p = pheromone[current][city]

                    p = max(p, 0.000001)

                    weight = (p ** alpha) * (
                        1 / (d + 0.001)
                    ) ** beta

                    choices.append(city)
                    weights.append(weight)

                city = random.choices(
                    choices, weights=weights, k=1
                )[0]

                route.append(city)
                unvisited.remove(city)

                if current == -1:
                    total += depot_dist[city]
                else:
                    total += dist[current][city]

                current = city

            total += depot_dist[current]
            routes.append(route)

        if total < best_distance:
            best_distance = total
            best_routes = [r[:] for r in routes]

        for i in range(n):
            for j in range(n):
                pheromone[i][j] = max(
                    pheromone[i][j] * (1 - rho),
                    0.000001
                )

        for i in range(n):
            depot_pheromone[i] = max(
                depot_pheromone[i] * (1 - rho),
                0.000001
            )

        for route in routes:

            first = route[0]
            depot_pheromone[first] += Q / total

            for i in range(len(route) - 1):
                a = route[i]
                b = route[i + 1]

                pheromone[a][b] += Q / total
                pheromone[b][a] += Q / total

            last = route[-1]
            depot_pheromone[last] += Q / total

    if (iteration + 1) % 20 == 0:
        print("Iteration", iteration + 1,
              "Best distance:", round(best_distance, 2))



print("\nTotal delivery distance:", round(best_distance, 2))
print("Delivery partners used:", partners)
print("Delivery locations covered:", n)
```
