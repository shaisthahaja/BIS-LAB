```python
import random
import math

start = (0, 0)
socket = (10, 0)

n = 10
iterations = 50
w = 0.5
c1 = 1.5
c2 = 1.5

def distance(a, b):
    return math.sqrt((a[0]-b[0])**2 + (a[1]-b[1])**2)

def fitness(p):
    return distance(start, p) + distance(p, socket)

pos = [[random.uniform(0, 10),
        random.uniform(-5, 5)] for i in range(n)]

vel = [[random.uniform(-1, 1),
        random.uniform(-1, 1)] for i in range(n)]

pbest = []

for p in pos:
    pbest.append(p.copy())

pbest_cost = [fitness(p) for p in pos]

best = 0

for i in range(1, len(pbest_cost)):
    if pbest_cost[i] < pbest_cost[best]:
        best = i

gbest = pbest[best].copy()

for step in range(iterations):
    for i in range(n):
        for j in range(2):
            r1 = random.random()
            r2 = random.random()

            vel[i][j] = (
                w * vel[i][j]
                + c1 * r1 * (pbest[i][j] - pos[i][j])
                + c2 * r2 * (gbest[j] - pos[i][j])
            )

            pos[i][j] += vel[i][j]

        cost = fitness(pos[i])

        if cost < pbest_cost[i]:
            pbest[i] = pos[i][:]
            pbest_cost[i] = cost

    best = pbest_cost.index(min(pbest_cost))
    gbest = pbest[best][:]

print("Starting point:", start)
print("Charging socket:", socket)
print("Best waypoint:", gbest)
print("Shortest path distance:", round(distance(start, socket), 2))
```
